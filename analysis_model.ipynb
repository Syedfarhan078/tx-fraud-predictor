Anomaly Detection Credit Card

Anomaly -> "something that is different from what is normal or usual"

We use Unsupervised learning algorithms that can automatically identify patterns in data without being explicitly told what to look for.

Two popular anomaly detection techniques are:

Autoencoders
Isolation forest
Autoencoders are a type of neural network that can learn to compress and reconstruct data.



Isolation forest is a tree-based forest algorithm that partitions data points and identifies anomalies based on their isolation scores.

Feature Engineering and Training a ML Model.

```python
df["isFraud"].value_counts()
```

```python
from sklearn.model_selection import train_test_split # for training and testing the data
from sklearn.preprocessing import StandardScaler # for scaling the data
from sklearn.linear_model import LogisticRegression # for modelling
from sklearn.metrics import classification_report, confusion_matrix # for model evaluation
from sklearn.pipeline import Pipeline # for training the model and tranforming the model together
from sklearn.compose import ColumnTransformer # data transformation
from sklearn.preprocessing import OneHotEncoder # data transformation
```

```python
df.head()
```

```python
# drop few coulums
df_model = df.drop(["nameOrig", "nameDest", "isFlaggedFraud"], axis = 1)
```

```python
df_model.head()
```

```python
# setting categorical and numerical types

categorical = ["type"]
numeric = ["amount", "oldbalanceOrg", "newbalanceOrig", "oldbalanceDest", "newbalanceDest"]
```

```python
# target (y) = isFraud
# X = dataframe (contains all columns except isFraud since it is target value)

y = df_model["isFraud"]
X = df_model.drop("isFraud", axis=1)
```

```python
# apply train, test, split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y)
```

```python
# preprocessiing the data

preprocessor = ColumnTransformer(
    transformers = [
        ("num", StandardScaler(), numeric),
        ("cat", OneHotEncoder(drop="first"), categorical)
    ],
    remainder = "drop"
)
```

```python
# creating model pipeline

pipeline = Pipeline([
    ("prep", preprocessor),
    ("clf", LogisticRegression(class_weight="balanced", max_iter=1000))
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
confusion_matrix(y_test, y_pred)
```

```python
# accuracy

pipeline.score(X_test, y_test) * 100
```

```python
import joblib
joblib.dump(pipeline, "fraud_detection_pipeline.pkl")
```
